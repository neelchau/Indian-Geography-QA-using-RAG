```python
import os
import threading
import subprocess
import time

def ollama_server():
    # Set the OLLAMA_HOST environment variable to listen on all interfaces
    os.environ['OLLAMA_HOST'] = '0.0.0.0:11434'
    # Start the Ollama server process
    subprocess.Popen(["ollama", "serve"])

# Start the server in a new thread
ollama_thread = threading.Thread(target=ollama_server)
ollama_thread.start()

# Give the server a moment to start
time.sleep(5)
print("Ollama server started.")
```

```python
from langchain_community.document_loaders import WikipediaLoader

query = [
    "Country - India(Bharat)",
    "Gujarat",
    "Delhi",
    "States and Union Territories of India",
    "Geography of India",
    "Geography of Indian States",
    "Geography of Gujarat"
]

docs = []
for q in query:
  loader = WikipediaLoader(query=q,load_max_docs=5)
  docs.extend(loader.load())
```

```python
print(len(docs))
```

```python
print(docs)
```

```python
for doc in docs:
  print(doc.metadata["title"])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 2000,
    chunk_overlap = 800,
)
```

```python
splits = text_splitter.split_documents(docs)
```

```python
print(len(splits))
```

```python
!ollama pull llama3
```

```python
from langchain.embeddings import OllamaEmbeddings
from langchain.vectorstores import Chroma

vectorstore = Chroma.from_documents(documents=splits, embedding=OllamaEmbeddings(base_url="http://localhost:11434",model="llama3"))
```

```python
print(vectorstore._collection.get())
```

```python
retriver = vectorstore.as_retriever()
```

```python
from langchain import hub
prompt = hub.pull("rlm/rag-prompt")
```

```python
from langchain_ollama import OllamaLLM

llm = OllamaLLM(model="llama3",base_url="http://localhost:11434")
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)
```

```python
rag_chain = (
    {
        "context": retriver | format_docs,
        "question": RunnablePassthrough()
    }
    | prompt
    | llm
)
```

```python
rag_chain.invoke("What is India?")
```

```python
rag_chain.invoke("How many States are in India")
```

```python
rag_chain.invoke("Tell me about Gujarat State")
```

```python
questions = [
    "List all States of India",
    "What is National Capital of India",
    "What is Gujarat Famous for?",
    "What is Ahmedabad?",
    "History of Gujarat"
]
```

```python
for question in questions:
  print(rag_chain.invoke(question))
  print("\n\n==============================================================\n\n")
```
